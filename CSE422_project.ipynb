# Handling Null Values

```python
lst=['Marital status','Application mode','Application order','Course','Daytime_evening_attendance','Previous qualification',"Mother's qualification",
     "Father's qualification","Mother's occupation","Displaced","Educational special needs","Debtor","Tuition fees up to date","Gender","Scholarship holder","International"
     ,'Nationality']
for col in lst:
    impute = SimpleImputer(missing_values=np.nan, strategy='most_frequent')

    impute.fit(academics[[col]])

    academics[col] = impute.transform(academics[[col]])
```

```python
lst1=['Previous qualification (grade)','Admission grade','International','Unemployment rate','Inflation rate','GDP']
for col in lst1:
    impute = SimpleImputer(missing_values=np.nan, strategy='mean')
    impute.fit(academics[[col]])
    academics[col] = impute.transform(academics[[col]])
```

```python
import numpy as np
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
from sklearn.multiclass import OneVsRestClassifier
```

```python
models = {
    "KNN": KNeighborsClassifier(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Neural Network": MLPClassifier(hidden_layer_sizes=(100,), activation='relu', solver='adam', max_iter=1000, random_state=1)
}
```

# Accuracy Using Standard Scaler

```python
# Train and evaluate models
accuracies = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    accuracies[name] = acc

# Plot bar chart
plt.figure(figsize=(10, 6))
bars = plt.bar(accuracies.keys(), accuracies.values(), color=['blue', 'green', 'orange', 'purple', 'red', 'cyan'])

# Add accuracy text on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 0.02, f'{height:.2f}', ha='center', va='bottom', fontsize=12)

plt.title("Model Accuracy Comparison ")
plt.ylabel("Accuracy")
plt.ylim(0, 1.1)  # extra space for labels
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

# Precisions Using Standard Scaler

```python
# Train and evaluate precision
precisions = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    prec = precision_score(y_test, y_pred, average='macro')  # 'macro' = average across classes
    precisions[name] = prec

# Plot precision bar chart
plt.figure(figsize=(10, 6))
bars = plt.bar(precisions.keys(), precisions.values(), color=['blue', 'green', 'orange', 'purple'])

# Add precision values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 0.02, f'{height:.2f}', ha='center', va='bottom', fontsize=12)

plt.title("Model Precision Comparison")
plt.ylabel("Precision (Macro Average)")
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

# Recalls Using Standard Scaler

```python
# Train and evaluate recall
recalls = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    rec = recall_score(y_test, y_pred, average='macro')  # 'macro' = average across classes
    recalls[name] = rec

# Plot recall bar chart
plt.figure(figsize=(10, 6))
bars = plt.bar(recalls.keys(), recalls.values(), color=['blue', 'green', 'orange', 'purple'])

# Add recall values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 0.02, f'{height:.2f}', ha='center', va='bottom', fontsize=12)

plt.title("Model Recall Comparison")
plt.ylabel("Recall (Macro Average)")
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

# F1 Score Using Standard Scaler

```python
# Train and evaluate F1 scores
f1_scores = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    f1 = f1_score(y_test, y_pred, average='macro')  # 'macro' = average across classes
    f1_scores[name] = f1

# Plot F1 score bar chart
plt.figure(figsize=(10, 6))
bars = plt.bar(f1_scores.keys(), f1_scores.values(), color=['blue', 'green', 'orange', 'purple'])

# Add F1 score values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, height + 0.02, f'{height:.2f}', ha='center', va='bottom', fontsize=12)

plt.title("Model F1 Score Comparison")
plt.ylabel("F1 Score (Macro Average)")
plt.ylim(0, 1.1)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

# Confusion Matrix Using Standard Scaler

```python
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm)

    plt.figure(figsize=(6, 4))
    disp.plot(cmap='Blues', values_format='d')
    plt.title(f"Confusion Matrix: {name}")
    plt.grid(False)
    plt.show()
```

```python
import numpy as np
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
from sklearn.multiclass import OneVsRestClassifier
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 8))

for name, model in models.items():
    # Fit using OneVsRestClassifier for multi-class ROC
    classifier = OneVsRestClassifier(model)
    classifier.fit(X_train, y_train)

    # Predict probabilities
    y_probs = classifier.predict_proba(X_test)

    # Check for NaN in y_probs and replace with 0 if found
    y_probs = np.nan_to_num(y_probs)

    # Binarize the labels
    y_test_bin = label_binarize(y_test, classes=np.unique(y_test))
    n_classes = y_test_bin.shape[1]

    # Compute ROC curve and ROC area for each class
    fpr = dict()
    tpr = dict()
    roc_auc = dict()
    for i in range(n_classes):
        fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_probs[:, i])
        roc_auc[i] = auc(fpr[i], tpr[i])

    # Compute micro-average ROC curve and ROC area
    fpr["micro"], tpr["micro"], _ = roc_curve(y_test_bin.ravel(), y_probs.ravel())
    roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])

    # Plot the micro-average ROC curve
    plt.plot(fpr["micro"], tpr["micro"],
             label=f'{name} (micro-average AUC = {roc_auc["micro"]:.2f})')


# Plot random line
plt.plot([0, 1], [0, 1], 'k--', label='Random Guess')

# Final plot setup
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison - Academic Success Dataset")
plt.legend(loc="lower right")
plt.grid()
plt.show()
```
